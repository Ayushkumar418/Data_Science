```python
import requests
```

```python
a = requests.get("https://quotes.toscrape.com/")
# print(a.text)
print(a.content)
```

```python
cwh = requests.get("https://codewithharry.com/")
print(cwh.text)
```

```python
wp = requests.get("https://www.wikipedia.org/")
print(wp.text)
```

```python
with open("HTML/page.html", "w", encoding="utf-8") as f:
    f.write(a.text)
    
with open("HTML/cwh.html", "w", encoding="utf-8") as f:
    f.write(cwh.text)
```

```python
print(a.status_code)
print(cwh.status_code)
print(wp.status_code)
```

```python
for i in range(1,51):
    url = f"https://books.toscrape.com/catalogue/page-{i}.html"
    code = requests.get(url)
    with open(f"HTML/page_{i}.html", "w", encoding="utf-8") as f:
        f.write(code.text)
    print(f"Page {i} successfully downloaded.")
```
